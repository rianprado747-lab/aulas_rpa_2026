```python
display(df['RA'].unique())
df['RA'] = df['RA'].astype(str).str.replace(r'\D', '', regex=True)
display(df['RA'].unique())
```

```python
colunas_notas = ['Nota01', 'TF01','Nota02','TF02']

for c in colunas_notas:
    df[c] = df[c].astype(str)

    df[c] = df[c].str.replace(',', '.')

    erros = ['FALTOU', 'falta', 'N/A', 'nan', '-', ' ' ]
    for p in erros :
        df[c] = df[c].str.replace(p,'0.0', regex=False)

    df[c] = pd.to_numeric(df[c], errors='coerce').fillna(0.0)

display(df.head())
display(df.info())
```
